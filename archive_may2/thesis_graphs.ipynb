```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Data setup
data = {
    'Precision': [0.95, 0.94, 0.94, 0.95],
    'Recall': [0.98, 0.86, 0.92, 0.95],
    'F1-Score': [0.96, 0.90, 0.93, 0.95]
}
index = ['No Nodule', 'Nodule', 'Macro Avg', 'Weighted Avg']
df = pd.DataFrame(data, index=index)

# Plotting the heatmap
plt.figure(figsize=(8, 5))
ax = sns.heatmap(df, annot=True, cmap="YlGnBu", vmin=0.8, vmax=1.0, fmt=".2f", linewidths=.5)
plt.title('Baseline ResNet-50 Classification Performance')
plt.yticks(rotation=0) 
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Data from the second (improved) classification report
data = {
    'Precision': [0.97, 0.90, 0.94, 0.95],
    'Recall': [0.96, 0.93, 0.95, 0.95],
    'F1-Score': [0.97, 0.91, 0.94, 0.95]
}
index = ['No Nodule', 'Nodule', 'Macro Avg', 'Weighted Avg']
df = pd.DataFrame(data, index=index)

plt.figure(figsize=(8, 5))

# Using the EXACT same visual scale as the first graph
ax = sns.heatmap(df, annot=True, cmap="YlGnBu", vmin=0.8, vmax=1.0, fmt=".2f", linewidths=.5)

plt.title('ResNet-50 w/ Multi-Scale Light Transformers Classification Performance')
plt.yticks(rotation=0) 
plt.tight_layout()
plt.show()
```
